```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Load the data
df = pd.read_csv(r'C:\Users\Slim 5pro 9LIN\Downloads\visual studio\customer-churn-project\data\raw\WA_Fn-UseC_-Telco-Customer-Churn.csv')

print("✅ Data loaded successfully!")
print(f"Dataset Shape: {df.shape}")
```

```python
# Check data quality issues
print("DATA QUALITY CHECK")
print("="*50)

# 1. Check data types
print("\n1. Data Types:")
print(df.dtypes)

# 2. Check for missing values
print("\n2. Missing Values Count:")
missing = df.isnull().sum()
print(missing[missing > 0])

# 3. Check for duplicate rows
print(f"\n3. Duplicate Rows: {df.duplicated().sum()}")

# 4. Check TotalCharges column (known issue)
print("\n4. TotalCharges Data Type Issue:")
print(f"   Current type: {df['TotalCharges'].dtype}")
print(f"   Sample values: {df['TotalCharges'].head()}")

# Try to convert to numeric and see errors
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print(f"   After conversion: {df['TotalCharges'].dtype}")
print(f"   New missing values: {df['TotalCharges'].isnull().sum()}")
```

```python
# Investigate missing values in TotalCharges
print("INVESTIGATING MISSING VALUES")
print("="*50)

# Look at rows with missing TotalCharges
print("\nRows with missing TotalCharges:")
missing_rows = df[df['TotalCharges'].isnull()]
print(missing_rows[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']])

print(f"\nNumber of missing rows: {len(missing_rows)}")
print(f"\nObservation: Customers with tenure = 0 have missing TotalCharges")

# Strategy: Fill missing TotalCharges with MonthlyCharges (since tenure is 0)
print("\n" + "="*50)
print("FIXING MISSING VALUES")
print("="*50)

# Fill missing TotalCharges
df['TotalCharges'].fillna(df['MonthlyCharges'], inplace=True)

# Verify fix
print(f"\nMissing values after fix: {df['TotalCharges'].isnull().sum()}")
print("✅ Missing values handled successfully!")
```

```python
# Investigate missing values in TotalCharges
print("INVESTIGATING MISSING VALUES")
print("="*50)

# Look at rows with missing TotalCharges
print("\nRows with missing TotalCharges:")
missing_rows = df[df['TotalCharges'].isnull()]
print(missing_rows[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']])

print(f"\nNumber of missing rows: {len(missing_rows)}")
print(f"\nObservation: Customers with tenure = 0 have missing TotalCharges")

# Strategy: Fill missing TotalCharges with MonthlyCharges (since tenure is 0)
print("\n" + "="*50)
print("FIXING MISSING VALUES")
print("="*50)

# Fill missing TotalCharges - use assignment instead of inplace
df['TotalCharges'] = df['TotalCharges'].fillna(df['MonthlyCharges'])

# Verify fix
print(f"\nMissing values after fix: {df['TotalCharges'].isnull().sum()}")
print("✅ Missing values handled successfully!")

# Show the previously missing rows now filled
print("\nPreviously missing rows (now filled):")
print(df.loc[missing_rows.index, ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']])
```

```python
# Final verification
print("FINAL DATA QUALITY CHECK")
print("="*50)

# Check all missing values
print("\n1. Missing Values in All Columns:")
print(df.isnull().sum().sum())

# Check data types
print("\n2. Data Types After Cleaning:")
print(df.dtypes)

# Summary
print("\n" + "="*50)
print("CLEANING SUMMARY")
print("="*50)
print("✅ TotalCharges converted from string to numeric")
print("✅ 11 missing values filled with MonthlyCharges")
print("✅ No duplicate rows")
print(f"✅ Clean dataset ready: {df.shape[0]} rows, {df.shape[1]} columns")
```
